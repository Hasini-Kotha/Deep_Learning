```python
import os
import gc
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import tensorflow_datasets as tfds

print("--- 1. Dual-GPU Setup ---")
# Ensure memory growth is set properly on BOTH T4 GPUs to prevent crashes
physical_devices = tf.config.list_physical_devices('GPU')
if physical_devices:
    try:
        for gpu in physical_devices:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"✅ Found {len(physical_devices)} GPUs. Memory growth is set.")
    except RuntimeError as e:
        print(e)
else:
    print("⚠️ WARNING: No GPU found.")

# Create the strategy to use BOTH GPUs in parallel
strategy = tf.distribute.MirroredStrategy()
print(f"Number of devices in sync: {strategy.num_replicas_in_sync}")

# Adjust batch size for multiple GPUs (32 per GPU -> 64 total)
BATCH_SIZE_PER_REPLICA = 32
GLOBAL_BATCH_SIZE = BATCH_SIZE_PER_REPLICA * strategy.num_replicas_in_sync

print("\n--- 2. Dataset Loading ---")
dataset, info = tfds.load('oxford_iiit_pet', with_info=True, as_supervised=True)
train_val_ds = dataset['train']
num_classes = info.features['label'].num_classes
num_train_val = info.splits['train'].num_examples

def preprocess(image, label):
    image = tf.image.resize(image, (224, 224))
    image = tf.keras.applications.mobilenet_v2.preprocess_input(image)
    return image, label

train_val_ds = train_val_ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)

print("\n--- 3. Model Definition ---")
def build_model(initializer='glorot_uniform', use_dropout=False, dropout_rate=0.5, use_batchnorm=False):
    base_model = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3),
                                                   include_top=False,
                                                   weights='imagenet')
    base_model.trainable = False 

    inputs = tf.keras.Input(shape=(224, 224, 3))
    x = base_model(inputs, training=False)
    x = tf.keras.layers.GlobalAveragePooling2D()(x)

    if use_batchnorm:
        x = tf.keras.layers.BatchNormalization()(x)
    if use_dropout:
        x = tf.keras.layers.Dropout(dropout_rate)(x)

    outputs = tf.keras.layers.Dense(num_classes, activation='softmax',
                                    kernel_initializer=initializer)(x)

    model = tf.keras.Model(inputs, outputs)
    return model

print("\n--- 4. K-Fold Cross-Validation ---")
# Define the 4 configurations for your LaTeX table
configs_to_test = {
    'C1': {'initializer': 'glorot_uniform', 'use_dropout': True, 'dropout_rate': 0.5, 'use_batchnorm': False},
    'C2': {'initializer': 'glorot_uniform', 'use_dropout': False, 'use_batchnorm': True},
    'C3': {'initializer': 'he_normal', 'use_dropout': True, 'dropout_rate': 0.25, 'use_batchnorm': True},
    'C4': {'initializer': 'glorot_uniform', 'use_dropout': False, 'use_batchnorm': False}
}

num_folds = 5
fold_size = num_train_val // num_folds
final_results = {}

# Set epochs to 6 for a safe, fast Kaggle run
EPOCHS_TO_RUN = 8

for config_name, params in configs_to_test.items():
    print(f"\nEvaluating Configuration: {config_name}")
    cv_accuracies = []
    
    for i in range(num_folds):
        # 1. Clear GPU memory completely between folds
        tf.keras.backend.clear_session()
        
        # 2. Slice dataset purely via pointers (Memory Safe!)
        val_fold_ds = train_val_ds.skip(i * fold_size).take(fold_size)
        train_before = train_val_ds.take(i * fold_size)
        train_after = train_val_ds.skip((i + 1) * fold_size)
        train_fold_ds = train_before.concatenate(train_after)

        # 3. Batch and prefetch using the Global Batch Size for BOTH GPUs
        train_fold_prep = train_fold_ds.shuffle(500).batch(GLOBAL_BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
        val_fold_prep = val_fold_ds.batch(GLOBAL_BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

        # 4. Build and compile model INSIDE strategy scope for Multi-GPU
        with strategy.scope():
            model_cv = build_model(**params)
            model_cv.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

        # 5. Train for exactly 6 epochs
        hist = model_cv.fit(train_fold_prep, validation_data=val_fold_prep, epochs=EPOCHS_TO_RUN, verbose=0)

        acc = max(hist.history['val_accuracy'])
        cv_accuracies.append(acc)
        print(f"  Fold {i+1} Accuracy: {acc:.4f}")

        # 6. Clean up variables to ensure Garbage Collector wipes RAM
        del model_cv, hist, train_fold_prep, val_fold_prep, val_fold_ds, train_before, train_after, train_fold_ds
        gc.collect()

    mean_acc = np.mean(cv_accuracies)
    std_acc = np.std(cv_accuracies)
    final_results[config_name] = (mean_acc, std_acc)
    print(f"--> 5-Fold CV Accuracy for {config_name}: {mean_acc:.4f} ± {std_acc:.4f}")

print("\n--- 5. Final Plot ---")
# Plotting the results
names = list(final_results.keys())
means = [res[0] for res in final_results.values()]
stds = [res[1] for res in final_results.values()]

plt.figure(figsize=(10, 6))
plt.bar(names, means, yerr=stds, capsize=10, color=['skyblue', 'lightgreen', 'salmon', 'gold'])
plt.title('Plot 13: 5-Fold Cross-Validation Accuracy Comparison')
plt.ylabel('Mean Validation Accuracy')
plt.ylim([0, 1.0])
plt.show()
```
