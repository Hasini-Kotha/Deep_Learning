```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step Activation Function
# -----------------------------
def step(x):
    return 1 if x >= 0 else 0

# -----------------------------
# Plot Decision Boundary
# -----------------------------
def plot_boundary(X, y, weights, bias, title):
    plt.figure(figsize=(5,5))

    # Plot points
    for i in range(len(y)):
        if y[i] == 0:
            plt.scatter(X[i,0], X[i,1], color='red', s=100)
        else:
            plt.scatter(X[i,0], X[i,1], color='blue', s=100)

    # Decision Boundary
    x_vals = np.linspace(-0.5,1.5,100)

    if abs(weights[1]) > 1e-6:
        y_vals = -(weights[0]*x_vals + bias)/weights[1]
        plt.plot(x_vals,y_vals,'k')

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)
    plt.grid(True)
    plt.title(title)
    plt.show()

# -----------------------------
# Perceptron Training
# -----------------------------
def train_gate(X, y, gate_name, lr=0.1, epochs=10):

    weights = np.zeros(2)
    bias = 0

    print("="*60)
    print(gate_name)
    print("="*60)

    update = 1

    for epoch in range(epochs):

        error_count = 0

        for i in range(len(X)):

            z = np.dot(X[i],weights)+bias
            pred = step(z)

            error = y[i]-pred

            if error != 0:

                weights += lr*error*X[i]
                bias += lr*error

                print(f"\nUpdate {update}")
                print("Input :",X[i])
                print("Target:",y[i])
                print("Prediction:",pred)
                print("Weights:",weights)
                print("Bias:",bias)

                plot_boundary(
                    X,y,
                    weights,
                    bias,
                    f"{gate_name} : Update {update}"
                )

                update += 1
                error_count += 1

        if error_count==0:
            break

    print("\nFinal Weights:",weights)
    print("Final Bias:",bias)

    print("\nPredictions")

    for sample in X:
        print(sample,"->",step(np.dot(sample,weights)+bias))

    plot_boundary(
        X,
        y,
        weights,
        bias,
        gate_name+" Final Decision Boundary"
    )
```

```python
# ==================================================
# AND Gate
# ==================================================

X_and=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_and=np.array([0,0,0,1])

train_gate(X_and,y_and,"AND Gate")
```

```python
# ==================================================
# OR Gate
# ==================================================

X_or=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y_or=np.array([0,1,1,1])

train_gate(X_or,y_or,"OR Gate")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------
# Step Activation Function
# ---------------------------------------
def step(x):
    return 1 if x >= 0 else 0


# ---------------------------------------
# Plot Decision Boundary
# ---------------------------------------
def plot_boundary(X, y, weights, bias, title):

    plt.figure(figsize=(5,5))

    # Plot data points
    for i in range(len(y)):
        if y[i] == 0:
            plt.scatter(X[i,0], X[i,1],
                        color='red', s=120,
                        label='Class 0' if i==0 else "")
        else:
            plt.scatter(X[i,0], X[i,1],
                        color='blue', s=120,
                        label='Class 1' if i==1 else "")

    x=np.linspace(-0.5,1.5,100)

    if abs(weights[1])>1e-6:
        y_line=-(weights[0]*x+bias)/weights[1]
        plt.plot(x,y_line,'k-',label='Decision Boundary')

    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)
    plt.grid(True)
    plt.legend()
    plt.title(title)
    plt.show()


# ---------------------------------------
# XOR Dataset
# ---------------------------------------

X=np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y=np.array([
    0,
    1,
    1,
    0
])


# ---------------------------------------
# Initialize
# ---------------------------------------

weights=np.zeros(2)

bias=0

learning_rate=0.1

epochs=10

update=1


print("="*60)
print("TRAINING SINGLE LAYER PERCEPTRON FOR XOR GATE")
print("="*60)


# ---------------------------------------
# Training
# ---------------------------------------

for epoch in range(epochs):

    print(f"\nEpoch {epoch+1}")

    errors=0

    for i in range(len(X)):

        weighted_sum=np.dot(X[i],weights)+bias

        prediction=step(weighted_sum)

        error=y[i]-prediction

        if error!=0:

            weights=weights+learning_rate*error*X[i]

            bias=bias+learning_rate*error

            errors+=1

            print("\nUpdate",update)
            print("Input      :",X[i])
            print("Target     :",y[i])
            print("Prediction :",prediction)
            print("Error      :",error)
            print("Weights    :",weights)
            print("Bias       :",bias)

            plot_boundary(
                X,
                y,
                weights,
                bias,
                f"Epoch {epoch+1} - Update {update}"
            )

            update+=1

    print("Errors in Epoch:",errors)


print("\n"+"="*60)
print("Final Weights :",weights)
print("Final Bias    :",bias)

print("\nFinal Predictions")

for sample in X:

    pred=step(np.dot(sample,weights)+bias)

    print(sample," -> ",pred)


plot_boundary(
    X,
    y,
    weights,
    bias,
    "Final Decision Boundary"
)

print("\n"+"="*60)
print("OBSERVATION")
print("="*60)

print("""
The perceptron fails to converge for the XOR gate because
the XOR dataset is NOT linearly separable.

No single straight line can separate the two classes.

Hence, the weights keep changing during training and
the algorithm cannot correctly classify all four samples.

This demonstrates the limitation of a Single Layer
Perceptron.
""")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# XOR Dataset
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
], dtype=np.float32)

y = np.array([0,1,1,0], dtype=np.float32)

# Build MLP
model = Sequential([
    Dense(2, input_shape=(2,), activation='tanh', name='Hidden'),
    Dense(1, activation='sigmoid', name='Output')
])

model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.5),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

############################################################
# Function to plot nonlinear decision boundary
############################################################

def plot_boundary(model, epoch):

    xx, yy = np.meshgrid(
        np.linspace(-0.5,1.5,200),
        np.linspace(-0.5,1.5,200)
    )

    grid = np.c_[xx.ravel(), yy.ravel()]

    pred = model.predict(grid, verbose=0)
    pred = pred.reshape(xx.shape)

    plt.figure(figsize=(6,6))

    plt.contourf(xx, yy, pred,
                 levels=np.linspace(0,1,20),
                 alpha=0.6,
                 cmap="coolwarm")

    plt.contour(xx,yy,pred,
                levels=[0.5],
                colors='black',
                linewidths=2)

    for i in range(len(X)):
        if y[i]==0:
            plt.scatter(X[i,0],X[i,1],
                        color='red',
                        marker='o',
                        s=120)
        else:
            plt.scatter(X[i,0],X[i,1],
                        color='blue',
                        marker='^',
                        s=120)

    plt.title(f"Decision Boundary after Epoch {epoch}")
    plt.xlim(-0.5,1.5)
    plt.ylim(-0.5,1.5)
    plt.grid(True)
    plt.show()

############################################################
# Train epoch-by-epoch
############################################################

epochs = 50

for epoch in range(epochs):

    model.fit(X,y,
              epochs=1,
              verbose=0)

    print("\n"+"="*60)
    print("Epoch",epoch+1)

    weights = model.get_weights()

    print("\nHidden Layer Weights")
    print(weights[0])

    print("\nHidden Layer Bias")
    print(weights[1])

    print("\nOutput Layer Weights")
    print(weights[2])

    print("\nOutput Layer Bias")
    print(weights[3])

    # Plot after every epoch
    plot_boundary(model,epoch+1)

############################################################
# Final Results
############################################################

print("\nFinal Predictions")

pred = model.predict(X)

for x,p in zip(X,pred):
    print(f"{x} --> {int(p>0.5)} ({p[0]:.4f})")

loss,acc = model.evaluate(X,y,verbose=0)

print("\nFinal Accuracy:",acc)
```
